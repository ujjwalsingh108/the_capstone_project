# Introduction to SentenceTransformer Encoders & Embedding Models

An **Embedding Model** is a specialized transformer architecture designed to convert variable-length text (words, sentences, or paragraphs) into dense, fixed-size numerical vectors in a continuous semantic space. Unlike generative Large Language Models (such as GPT-4 or Gemini) that predict the next token, encoder models output dense mathematical representations where **geometric distance reflects semantic similarity**.

---

## The `all-MiniLM-L6-v2` Model

The `sentence-transformers/all-MiniLM-L6-v2` model is one of the most widely adopted lightweight embedding models in the open-source NLP ecosystem.

* **Architecture:** 6-layer BERT-style transformer encoder distilled from larger architectures.
* **Vector Dimensionality:** Maps any input text into a **384-dimensional dense vector**.
* **Sequence Length:** Supports context windows up to 256 tokens (optimal for sentences, paragraphs, and short product summaries).
* **Speed & Footprint:** Weighs only $\approx 80\text{ MB}$, allowing it to execute at ultra-low latency on standard CPUs without requiring dedicated GPUs.

---

## Comparison: Open-Source Local Embeddings vs. Proprietary API Embeddings

| Feature | Open-Source (`all-MiniLM-L6-v2`, `bge-small`) | Proprietary API (`text-embedding-3-small`, Google Vertex) |
| --- | --- | --- |
| **Cost** | **$0.00** (Free, unlimited inferences) | Pay-per-token billed on every index/query operation |
| **Latency** | Single-digit milliseconds locally (no network roundtrips) | Subject to internet latency and API rate limits ($50\text{–}200\text{ ms}$) |
| **Data Privacy** | **100% On-Premise**: Data never leaves local memory | Payloads are transmitted over the wire to third-party servers |
| **Dimensions** | 384 dimensions (minimal RAM/Vector DB storage) | 1536 to 3072 dimensions (higher storage & compute requirements) |
| **Air-Gapped Support** | Works completely offline | Requires persistent external internet access |

---

## Implementation Examples

### 1. Basic Semantic Similarity with `sentence-transformers`

```python
from sentence_transformers import SentenceTransformer, util

# 1. Load the model (downloads automatically on first run)
model = SentenceTransformer("all-MiniLM-L6-v2")

# 2. Define text corpus
corpus = [
    "Apple iPhone 15 Pro Max 256GB Natural Titanium",
    "Sony WH-1000XM5 Wireless Noise-Canceling Headphones",
    "Samsung Galaxy S24 Ultra Android Smartphone",
    "Men's Casual Slim Fit Cotton T-Shirt"
]

# 3. Compute 384-dimensional dense embeddings
corpus_embeddings = model.encode(corpus, convert_to_tensor=True)

# 4. Perform a semantic query search
query = "Looking for a high-end mobile phone"
query_embedding = model.encode(query, convert_to_tensor=True)

# Compute cosine similarity
similarities = util.cos_sim(query_embedding, corpus_embeddings)[0]

# Print ranked results
for idx in similarities.argsort(descending=True):
    print(f"Score: {similarities[idx]:.4f} | Product: {corpus[idx]}")

```

### 2. Price Estimation / Feature Vector Extraction for ML Classifiers

In price estimation pipelines, embeddings can be extracted from product text summaries and directly passed into tabular regression models (such as XGBoost, LightGBM, or Ridge Regression):

```python
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import Ridge

model = SentenceTransformer("all-MiniLM-L6-v2")

# Training summaries and known target prices
train_summaries = [
    "Budget wired gaming mouse with RGB lighting",
    "Flagship 4K OLED HDR Smart TV 65-inch",
    "Stainless steel insulated water bottle 1L",
    "Professional mirrorless camera with 4K video"
]
train_prices = np.array([25.00, 1899.00, 22.50, 2400.00])

# Extract embedding matrices (N samples x 384 features)
X_train = model.encode(train_summaries)

# Train downstream linear/ridge regression model
regressor = Ridge(alpha=1.0)
regressor.fit(X_train, train_prices)

# Predict price on a new, unseen product summary
new_product = ["Logitech wireless ergonomic computer mouse"]
new_embedding = model.encode(new_product)
predicted_price = regressor.predict(new_embedding)

print(f"Estimated Price: ${predicted_price[0]:.2f}")

```

## Alternative Open-Source Embedding Models

Depending on the scale and accuracy requirements of a project, other modern encoder alternatives include:

* **`BAAI/bge-small-en-v1.5`** (384 dims): Slightly higher retrieval benchmark performance than `all-MiniLM-L6-v2` with a similar footprint.
* **`intfloat/e5-small-v2`** (384 dims): Strong performance across passage ranking tasks requiring prefix queries (`query: ...` vs `passage: ...`).
* **`BAAI/bge-large-en-v1.5`** (1024 dims): Heavyweight model for maximum semantic retrieval precision when GPU acceleration is available.

````python
COLLECTION = "products_vectorstore"

EXISTING_COLLECTION_NAMES = [collection.name for collection in client.list_collections()]
print(EXISTING_COLLECTION_NAMES)
````

````python
# Target is to embed the corpus into the vector store collection, ids, raw documents and metadata

if COLLECTION not in EXISTING_COLLECTION_NAMES:
    print(f"Collection '{COLLECTION}' does not exist. Creating it.")
    client.create_collection(name=COLLECTION)

# Loop through the lite training dataset
for i in tqdm(range(0,len(train), 1000)):
    # collect summary of each item
    summary = [item.summary for item in train[i:i+1000]]
    # embed the summary into the vector store
    embeddings = model.encode(summary).tolist()
    # metadata for each item
    metadata = [{"category": item.category, "price": item.price} for item in train[i:i+1000]]
    # create unique ids for each item
    ids = [f"id_{j}" for j in range(i, i+1000)]

    # add the embeddings, ids, and metadata to the collection
    collection.add(
        ids=ids,
        embeddings=embeddings,
        metadatas=metadata,
        documents=summary
    )
````

# Type Conversions Between NumPy Arrays and Python Lists

In data engineering and AI retrieval pipelines, converting data between **Python `list**` and **NumPy `ndarray**` (`np.array`) is a standard pattern driven by the distinct requirements of **serialization/APIs** versus **vectorized mathematical computation**.

---

## 1. Why Convert to Python `list` When Adding to ChromaDB?

In your ingestion step:

```python
embeddings = model.encode(summary).tolist()

```

### Key Reasons:

* **JSON Serialization Compatibility:**
Vector databases (like ChromaDB) communicate internally or over the network via standard JSON-compatible formats or lightweight binary protocols. Native NumPy numeric types (e.g., `np.float32`, `np.float64`) are not directly JSON-serializable in standard Python and will trigger serialization errors (`TypeError: Object of type float32 is not JSON serializable`).
* **Database Driver API Requirements:**
The ChromaDB Python client expects standard Python collection primitives (`list[list[float]]` for dense vectors, `list[str]` for documents/IDs, `list[dict]` for metadata). `.tolist()` converts the underlying C-contiguous block of floats into native Python float objects.

---

## 2. Why Convert Back to `np.array` for Dimensionality Reduction (t-SNE / PCA / UMAP)?

In your visualization step:

```python
reduced_vectors = tsne.fit_transform(np.array(result["embeddings"]))

```

### Key Reasons:

* **Vectorized Mathematical Operations:**
Algorithms like t-SNE, PCA, and UMAP (from `scikit-learn`) perform matrix multiplications, pairwise Euclidean distance calculations, and gradient updates. These routines run on optimized C/Fortran and BLAS/LAPACK backends, which require a contiguous 2D memory block of shape `(N, D)`.
* **2D Matrix Slicing:**
Pure Python nested lists do not support multidimensional slicing (e.g., `matrix[:, 0]`). `np.array()` wraps the data so you can cleanly index dimensions for plotting:
```python
x = reduced_vectors[:, 0]  # First component (X-axis)
y = reduced_vectors[:, 1]  # Second component (Y-axis)

```


* **Type Validation in `scikit-learn`:**
Methods like `tsne.fit_transform()` expect array-like structures that provide attributes like `.shape`, `.dtype`, and `.ndim`. While `scikit-learn` can internally cast nested lists, passing an explicit `np.array` prevents unexpected memory duplications or type ambiguity warnings.

---

## Summary of the Data Flow

| Stage | Data Format | Reason |
| --- | --- | --- |
| **Model Output** (`model.encode`) | `np.ndarray` or `torch.Tensor` | Output of the transformer neural network matrix multiplications. |
| **Ingestion** (`collection.add`) | `list[list[float]]` (`.tolist()`) | Required for JSON/RPC schema validation and database insertion. |
| **Database Retrieval** (`collection.get`) | `list[list[float]]` | Deserialized raw Python lists returned from the database index. |
| **Manifold Learning** (`tsne.fit_transform`) | `np.ndarray` (`np.array(...)`) | Matrix mathematics, pairwise distance kernels, and 2D slicing. |

````python
def find_similar_items(item: Item):
    item_vector = vectorize_item(item)
    results = collection.query(
        query_embeddings = item_vector.astype('float32').tolist(),
        n_results = 5
    )
    documents = results['documents'][0]
    prices = [i["price"] for i in results['metadatas'][0]]

    return documents, prices
````

````python
find_similar_items(test[0])
````

````python
# We need to give some context to the frontier modal by selecting 5 products with similar description

def make_context_for_frontier(item: Item):
    documents, prices = find_similar_items(item)
    context = ""
    for doc, price in zip(documents, prices):
        context += f"Description: {doc}\nPrice: {price}\n\n"
    return context
````

````python
print(make_context_for_frontier(test[0]))
````

````python
def frame_message_for_frontier(item: Item):
    context = make_context_for_frontier(item)
    message = f"Estimate the price of this product. Respond with the price, no explanation\n\n{item.summary}\n\n"
    message += f"Context:\n{context}"
    return [
        {
            "role": "user",
            "content": message
        }
    ]
````

````python
print(frame_message_for_frontier(test[0]))
````

````python
def groq_query_for_frontier(item: Item):
    response = completion(
        model = "groq/openai/gpt-oss-20b",
        messages = frame_message_for_frontier(item),
        reasoning_effort = "high",
        seed = 42,
        allowed_openai_params = ["reasoning_effort"],
    )
    return response.choices[0].message.content
````
